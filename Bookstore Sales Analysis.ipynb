```python
import pandas as pd
from IPython.display import display
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
# Load Dataset
df = pd.read_csv("bookstore_sales.csv")
print(df.head())
```

```python
# periksa info termasuk tipe data
df.info()
```

```python
# cleaning data
df.isnull().sum()
df.drop('dataset', axis=1, inplace=True)
df.drop('year', axis=1, inplace=True)
df.drop('month', axis=1, inplace=True)
```

```python
# pemformatan data
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['revenue'] = df['revenue'].astype(int)
df['net_revenue'] = df['net_revenue'].astype(int)
```

```python
df.info()
```

```python
# Total Transaksi
total_trans = df['transactions'].sum()
# Total Pendapatan kotor
total_revenue = df['revenue'].sum()
# Total Pendapatan bersih
total_net_revenue = df['net_revenue'].sum()

display(f'Total Transaksi : {total_trans}')
display(f'Total Pendapatan : {total_revenue}')

# Rata-Rata Pendapatan
mean_revenue = df['revenue'].mean()
display (f'Rata Rata Pendapatan : {mean_revenue}')
#Rata-Rata Transaksi
mean_trans = df['transactions'].mean()
display(f'Total Transaksi : {mean_trans}')
```

```python
# Peringkat Toko berdasarkan penjualan
df.groupby('store_id')['revenue'].sum().sort_values(ascending=False)
```

```python
# Peringkat Channel penjualan Terlaris
df.groupby('channel')['revenue'].sum().sort_values(ascending=False)
```

```python
# Cek korelasi antar variabel sebelum dilakukan group by
print(df[['transactions', 'revenue', 'gst_collected', 'net_revenue']].corr())
```

```python
#kategorisasi untuk mengetahui tren data
print(df.groupby('year')[['transactions', 'revenue', 'gst_collected', 'net_revenue']].sum())
```

```python
print(df.groupby('channel')[['transactions', 'revenue', 'gst_collected', 'net_revenue']].sum())
```

```python
df_pivot = df.pivot_table(
    index='store_id',
    columns='channel',
    values='net_revenue',
    aggfunc='sum',
    fill_value=0
)

print(df_pivot)
```

```python
# Expor Data Bersih
df.to_csv('clean_bookstore_sales.csv', index=False)
```
