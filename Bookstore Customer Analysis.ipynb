```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
#load Dataset
df = pd.read_csv("bookstore_customers_expanded.csv")
df.info()
```

```python
df.head(10)
```

```python
#Mengganti data NULL/NAN dengan keterangan
df['escalation_reason']= df['escalation_reason'].fillna('No Description')
```

```python
# Jumlah Customer berdasarkan ID unik, supaya tidak duplikat
display (f"Total customer : {df['customer_id'].nunique()}")

#rata-rata total spent
display(f'Average of Total Spent: {df['total_spent'].mean()}')

#rata-rata anual visits
display(f'Average of Annual Visits : {df['annual_visits'].mean()}')

# rata-rata online ratio
display(f'Average of Online Ration : {df['online_ratio'].mean()}')

# rata rata ltv
display(f'Average of ltv : {df['ltv_estimate'].mean()}')

# distribusi ratting
display(f'Average of Customer Rating : {df['customer_rating'].mean()}')
```

```python
# Toko teratas berdasarkan jumlah customer
df['home_store'].value_counts()
```

```python
# Region customer teratas
df['region'].value_counts()
```

```python
#Melihat dominasi customer online
online_dom = (df['online_ratio'] > 0.5).mean() * 100
online_dom
```

```python
#Melihat dominasi customer offline
offline_dom = (df['online_ratio'] < 0.5).mean() * 100
offline_dom
```

```python
#Melihat dominasi customer hybrid (online dan offline)
Balanced = (df['online_ratio'] == 0.5).mean() * 100
Balanced
```

```python
# menyatukan dalam bentuk dataframe dictionary
pd.DataFrame({
    'Category': ['Online-Dominant', 'Offline-Dominant', 'Balanced'],
    'Percentage': [round(online_dom, 2), round(offline_dom, 2), round(Balanced, 2)]
})
```

```python
# jumlah pengguna loyalty vs non
df['loyalty_member'].value_counts()
```

```python
# persentase perbandingan pengguna loyalty dan non
percentage = df['loyalty_member'].value_counts(normalize=True) * 100
round(percentage, 2)
```

```python
#Rata rata kunjungan tahun 2023
df['visits_2023'].mean()
```

```python
#Jumlah kunjungan tahun 2023
df['visits_2023'].sum()
```

```python
df['visits_2023'].describe()
```

```python
#Pelabelan/klasifikasi berdasarkan jumlah kunjungan
df['visit_category'] = pd.cut(
    df['visits_2023'],
    bins=[0, 2, 6, 20, 100],
    labels=['Sangat Jarang', 'Jarang', 'Sering', 'Sangat Sering']
)

df['visit_category'].value_counts()
```

```python
# top customer yang merupakan spender tertinggi
top10_spenders = df.nlargest(10, 'total_spent')[['customer_id','total_spent','total_purchases','home_store','region']]
display(top10_spenders)
```

```python
#Ringkasan statistik
df['total_spent'].describe()        # summary stats
```

```python
# untuk melihat distribusi kasar
spending_bins = pd.cut(df['total_spent'], bins=[-1,50,100,250,500,1000,5000, np.inf])
df.groupby(spending_bins, observed=False)['customer_id'].count()
```

```python
#Rata Rata pengeluaran pelanggan
avg_spent_by_region = df.groupby('region')['total_spent'].mean().sort_values(ascending=False)
display(avg_spent_by_region)
```

```python
#Rata-rata pengeluaran loyalty member vs non loyalty
avg_spent_loyalty = df.groupby('loyalty_member')['total_spent'].agg(['count','mean','median']).reset_index()
display(avg_spent_loyalty)
```

```python
# korelasi online_ratio, total_spent, annual_visits

for col in ['online_ratio','total_spent','annual_visits','total_purchases','customer_rating']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Pearson correlations
corr_online_total = df['online_ratio'].corr(df['total_spent'])
corr_visits_purchases = df['annual_visits'].corr(df['total_purchases'])
corr_rating_spent = df['customer_rating'].corr(df['total_spent'])

corrs = {
    'online_ratio_vs_total_spent': corr_online_total,
    'annual_visits_vs_total_purchases': corr_visits_purchases,
    'customer_rating_vs_total_spent': corr_rating_spent
}
corrs_df = pd.DataFrame({
    'Metric': [
        'Online Ratio vs Total Spent',
        'Annual Visits vs Total Purchases',
        'Customer Rating vs Total Spent'
    ],
    'Correlation': [
        corr_online_total,
        corr_visits_purchases,
        corr_rating_spent
    ]
})

corrs_df
```

```python
df.info()
```

```python
# karena kolom last_purchase_date dan first_purchase_date masih bertipe object maka ubah ke datetime

df['last_purchase_date'] = pd.to_datetime(df['last_purchase_date'], errors='coerce')
df['first_purchase_date'] = pd.to_datetime(df['first_purchase_date'], errors='coerce')
```

```python
# customer aktif transaksi dalam 90 hari terakhir
today = df['last_purchase_date'].max()
active_90_mask = (today - df['last_purchase_date']).dt.days <= 90
active_90_count = active_90_mask.sum()
active_90_pct = active_90_mask.mean() * 100
active_90_count, active_90_pct
```

```python
# Menghitung Jumlah Customer Dorman(tidak bertransaksi lebih dari 365 hari terakhir)
dormant_mask = (today - df['last_purchase_date']).dt.days > 365
dormant_count = dormant_mask.sum()
dormant_pct = dormant_mask.mean() * 100
dormant_count, dormant_pct
```

```python
#melihat jumlah kunjungan 90 hari terakhir
df['recency_days'] = (today - df['last_purchase_date']).dt.days
df['lifetime_days'] = (df['last_purchase_date'] - df['first_purchase_date']).dt.days
df[['recency_days','lifetime_days']].describe()
```

```python
# Melihat jumlah customer yang pernah komplain
escalation_counts = df['has_escalation'].value_counts()
escalation_pct = df['has_escalation'].value_counts(normalize=True) * 100
escalation_counts, escalation_pct
```

```python
# Alasan Complain Teratas
top_escalation_reasons = df.loc[df['has_escalation']==True, 'escalation_reason'].value_counts().head(10)
display(top_escalation_reasons)
```

```python
# Melihat Dampak komplain terhadap total_spent, customer_rating, loyalty_member

impact = df.groupby('has_escalation').agg({
    'total_spent': ['count','mean','median'],
    'customer_rating': ['mean','median'],
    'loyalty_member': lambda x: x.mean()  # proporsi yang loyal
})
# beri nama kolom yang rapi
impact.columns = ['_'.join(col).strip() for col in impact.columns.values]
display(impact)
```
